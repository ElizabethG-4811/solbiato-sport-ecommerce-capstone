```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import re
import glob
import os
import time
```

```python
#Direct Feed Url 
url = "https://solbiatosport.com/product-category/solbiato-kids/feed/"
```

```python
page_number = 1
domain = "https://solbiatosport.com/"
path = f"product-category/solbiato-kids/feed/?paged={page_number}"
url = domain + path

headers = {'User-Agent': 'Mozilla/5.0 (Macintosh; Intel Mac OS X 10_15_7)'}

print(f"🚀 Pulling Kids data from: {url}")
response = requests.get(url, headers=headers)

if response.status_code == 200:
    soup = BeautifulSoup(response.content, 'xml')
    items = soup.find_all('item')
    
    kids_products = []
    for item in items:
        name = item.find('title').text
        link = item.find('link').text
        
        # Pulling the raw description which contains the price and material
        desc = item.find('description').text if item.find('description') else "N/A"
        
        # --- NEW: Cleaning the price ---
        # This looks for a '$' followed by digits and decimals
        price_match = re.search(r'\$\d+(?:\.\d+)?', desc)
        clean_price = price_match.group(0) if price_match else "N/A"
        
        kids_products.append({
            'product_name': name,
            'category': 'Kids',
            'price': clean_price,
            'description_raw': desc,
            'url': link
        })
    
    # Create the DataFrame
    df_kids = pd.DataFrame(kids_products)
    print(f"✅ Success! Found {len(df_kids)} products on Page {page_number}.")
    
    # Display the first few rows to verify the clean prices
    display(df_kids)
    
    # Save to your raw folder - using the page number in the filename
    file_path = f'../data/raw/kids_page_{page_number}_raw.csv'
    df_kids.to_csv(file_path, index=False)
    print(f"💾 Saved to: {file_path}")

else:
    print(f"❌ Failed to reach site. Status Code: {response.status_code}")

```

```python
import glob
import os

#Merging all Solbiato Kids Files 

# 1. Be very specific about the path
# If you are in the 'notebooks' folder, we go up one level then into 'data/raw'
path = '../data/raw/'
all_files = glob.glob(os.path.join(path, "kids_page_*_raw.csv"))

li = []

print(f"Checking {len(all_files)} files...")

for filename in all_files:
    try:
        # We try to read it; if it's empty, it will jump to the 'except' part
        df = pd.read_csv(filename)
        if not df.empty:
            li.append(df)import pandas as pd
            # print(f"Added: {os.path.basename(filename)}")
    except pd.errors.EmptyDataError:
        print(f"Skipping empty file: {os.path.basename(filename)}")
    except Exception as e:
        print(f"Error reading {os.path.basename(filename)}: {e}")

# 2. The Final Stack
if li:
    df_kids_final = pd.concat(li, axis=0, ignore_index=True)
    df_kids_final.drop_duplicates(inplace=True)
    
    output_path = '../data/raw/solbiato_kids_COMPLETE.csv'
    df_kids_final.to_csv(output_path, index=False)
    
    print(f"\n✅ SUCCESS!")
    print(f"Merged {len(li)} files into {output_path}")
    print(f"Total Products: {len(df_kids_final)}")
else:
    print("\n❌ No valid data found in any of the files.")
```

```python
# 1. SETTINGS
category_name = "Sport-Tops" 
# This is the EXACT full URL with the 'feed' included
base_feed_url = "https://solbiatosport.com/product-category/solbiato-sport/sport-tops/feed/"

# 2. THE AUTOMATED LOOP
all_products = []
current_page = 1

while True:
    # We add ?paged= to the end of the full feed URL
    url = f"{base_feed_url}?paged={current_page}"
    print(f"🕵️ Scraping {category_name}: Page {current_page}...")
    
    headers = {'User-Agent': 'Mozilla/5.0'}
    response = requests.get(url, headers=headers)
    
    # Stop if the page doesn't exist
    if response.status_code != 200:
        print(f"🏁 Reached the end at Page {current_page - 1}.")
        break
        
    soup = BeautifulSoup(response.content, 'xml')
    items = soup.find_all('item')
    
    # Stop if no items are found
    if not items:
        print(f"🏁 No more products found after Page {current_page - 1}.")
        break
        
    for item in items:
        name = item.find('title').text
        link = item.find('link').text
        desc = item.find('description').text if item.find('description') else "N/A"
        
        all_products.append({
            'product_name': name,
            'category': category_name,
            'description_raw': desc,
            'url': link
        })
    
    current_page += 1
    time.sleep(1) # Be polite to the server!

# 3. SAVE THE MASTER FILE
df_tops_master = pd.DataFrame(all_products)
output_path = "../data/raw/solbiato_tops_COMPLETE.csv"
df_tops_master.to_csv(output_path, index=False)

print(f"\n✅ DONE! Collected {len(df_tops_master)} total Tops.")
display(df_tops_master.head())
```

```python
# --- THE SURGICAL PRICE SNATCHER ---
import requests
from bs4 import BeautifulSoup

# We'll use the K-Darak URL you just found
test_url = "https://solbiatosport.com/product/k-darak/"
print(f"🕵️ Target acquired: {test_url}")

headers = {'User-Agent': 'Mozilla/5.0'}
response = requests.get(test_url, headers=headers)
soup = BeautifulSoup(response.content, 'html.parser')

# 1. Look for the 'price' container
price_container = soup.find('p', class_='price')

if price_container:
    # 2. Find ALL price tags inside it
    all_prices = price_container.find_all('bdi')
    
    if len(all_prices) > 1:
        # If there are two prices, the LAST one is always the Sale Price
        final_price = all_prices[-1].get_text(strip=True)
        print(f"🔥 SALE PRICE FOUND: {final_price}")
    else:
        # If there's only one, that's the regular price
        final_price = all_prices[0].get_text(strip=True)
        print(f"💰 REGULAR PRICE: {final_price}")
else:
    print("❌ Price tag not found.")
```
